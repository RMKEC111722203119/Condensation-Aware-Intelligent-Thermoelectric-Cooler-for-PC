```python
param=5
```

```python
df1 = pd.read_csv(f"/content/drive/MyDrive/A Sen data/20 load/24/{param}v 20p.csv", encoding='latin1')
df1
```

```python
df2 = pd.read_csv(f"/content/drive/MyDrive/A Sys data/20 load/24cwithpeltier/{param}V 20p.csv", encoding='latin1')
df2
```

```python
import pandas as pd

# Ensure Timestamp is datetime in both dataframes
df1['Timestamp'] = pd.to_datetime(df1['Timestamp'])
df2['Timestamp'] = pd.to_datetime(df2['Timestamp'])

# Merge on Timestamp
merged = pd.merge(df1, df2, on='Timestamp', how='inner')
merged.drop(merged.index[:5], inplace=True)
# Optional: Check result
merged
```

```python
merged['Core1_Load_pct'] = merged['Core1_Load_pct'].replace(100.00, 20)
merged['Core2_Load_pct'] = merged['Core2_Load_pct'].replace(100.00, 20)
display(merged.head())
```

```python
if param==10:
  param=11
```

```python
merged['volt'] = param
display(merged.head())
```

```python
merged.to_csv(f"/content/drive/MyDrive/A Merged data/20 load/24/{param}v 20p",index=False)
```

```python
merged_sorted = merged.sort_values(by=['Core1_Temp_C', 'Core2_Temp_C','Humidity (%)'], ascending=[ True, True,False])
display(merged_sorted)
```

```python
merged_sorted.to_csv(f"/content/drive/MyDrive/A sorted data/20 load/24/{param}v 20p",index=False)
```
